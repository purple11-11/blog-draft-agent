```python
import os

from dotenv import load_dotenv, find_dotenv

from pathlib import Path
from bs4 import BeautifulSoup

from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings

load_dotenv(find_dotenv(usecwd=True))
if not os.environ.get("OPENAI_API_KEY"):
    raise SystemExit("agentic-ai 폴더의 .env 파일에 OPENAI_API_KEY 한 줄을 넣습니다.")

CORPUS_PATH = Path("corpus_raw")

docs = []
for post_dir in CORPUS_PATH.iterdir():
    if not post_dir.is_dir(): # 폴더가 아닌 게 섞여 있을 수 있으니 .is_dir()로 걸러내기
        continue
    file = open(f"{post_dir}/index.html", encoding="utf-8-sig",)
    soup = BeautifulSoup(file,"html.parser")
    title = soup.find(class_="se-title-text").get_text(strip=True)

    paragraphs = soup.find_all(class_="se-text-paragraph")
    body = "\n".join(p.get_text(strip=True) for p in paragraphs)

    text = title + "\n\n" + body
    docs.append(Document(page_content=text, metadata={"folder": post_dir.name}))




emb = OpenAIEmbeddings(model="text-embedding-3-small")

db = Chroma.from_documents(
    docs, emb,
    persist_directory="chroma_db",
    ids=[f"folder-{d.metadata['folder']}" for d in docs]
)

def retrieve(query:str, k:int = 2):
    return db.similarity_search_with_score(query, k)

print(retrieve("전주 맛집", 2))
```
